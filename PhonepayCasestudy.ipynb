```python
#Q3.1 Data Quality Checks
#[Verify that district-level data adds up correctly to state-level data]
```

```python
#Aggregate District-Level Data (State-wise)
#We sum district data to state level.
district_state_sum = district_txn_users.groupby("State").agg(
District_Transactions = ("Transactions", "sum"),
District_Amount = ("Amount (INR)", "sum"),
District_Users = ("Registered Users", "sum")).reset_index()

district_state_sum.head()
```

```python
#Aggregate State-Level Data
#Now calculate totals from state dataset.
state_sum = state_txn_users.groupby("State").agg(
State_Transactions = ("Transactions", "sum"),
State_Amount = ("Amount (INR)", "sum"),
State_Users = ("Registered Users", "sum")
).reset_index()

state_sum.head()
```

```python
#Merge District & State Data for Comparison
comparison = pd.merge(state_sum,district_state_sum,on="State",
    how="inner"
)
comparison.head()
```

```python
#Calculate Differences (Very Important)
comparison["Txn_Diff"] = comparison["State_Transactions"] - comparison["District_Transactions"]
comparison["Amount_Diff"] = comparison["State_Amount"] - comparison["District_Amount"]
comparison["User_Diff"] = comparison["State_Users"] - comparison["District_Users"]

comparison.head()
```

```python
#Q4.1 Ratio of Users to Population (State-wise)
#Step 1: Get total population per state
state_population = district_demo.groupby("State")["Population"].sum().reset_index()
state_population.head(3)
```

```python
#Get total registered users per state
state_users = state_txn_users.groupby("State")["Registered Users"].sum().reset_index()
state_users.head()
```

```python
#Merge both datasets
users_population = pd.merge(state_users,state_population,
    on="State"
)
users_population.head()
```

```python
#Calculate ratio
users_population["User_to_Population_Ratio"] = (
users_population["Registered Users"] / users_population["Population"]
)

users_population.head()
```

```python
#Q4.1.2 Column Charts
import matplotlib.pyplot as plt
plt.figure()
plt.bar(users_population["State"], users_population["User_to_Population_Ratio"])
plt.xticks(rotation=90)
plt.title("User to Population Ratio by State")
plt.ylabel("Ratio")
plt.tight_layout()
plt.show()
```

```python
#Q4.2Correlation: Population Density vs Transaction Volume
#Step 1: Merge district datasets
district_merge = pd.merge(district_txn_users,district_demo,
    on=["State", "District", "Code"]
)
district_merge.head()
```

```python
#Step 2: Calculate correlation
correlation = district_merge["Density"].corr(district_merge["Transactions"])
correlation
```

```python
plt.figure()
plt.scatter(district_merge["Density"], district_merge["Transactions"])
plt.xlabel("Population Density")
plt.ylabel("Transaction Volume")
plt.title("Population Density vs Transactions")
plt.tight_layout()
plt.show()
```

```python
#4.3 Average Transaction Amount per User (State-wise)
# Step 1: Aggregate total amount & users
state_amount_users = state_txn_users.groupby("State").agg(
Total_Amount=("Amount (INR)", "sum"),
Total_Users=("Registered Users", "sum")
).reset_index()
state_amount_users.head()
```

```python
#Calculate average per user
state_amount_users["Avg_Amount_Per_User"] = (
state_amount_users["Total_Amount"] / state_amount_users["Total_Users"]
)

state_amount_users.head()
```

```python
#Top 5 States
state_amount_users.sort_values(
    "Avg_Amount_Per_User", ascending=False).head(5)
```

```python
#Bottom 5 States
state_amount_users.sort_values(
    "Avg_Amount_Per_User").head(5)
```

```python
#Q4.4Device Brand Usage Ratio
#Step 1: Total users per state
total_users_state = state_txn_users.groupby("State")["Registered Users"].sum().reset_index()
total_users_state.head()
```

```python
#Step 2: Merge with device data
device_merge = pd.merge(
    state_device,
    total_users_state,
    on="State",
    suffixes=("_Brand", "_Total")
)
device_merge.head()
```

```python
#Step 3: Calculate device usage ratio
device_merge["Device_Usage_Ratio"] = (
    device_merge["Registered Users_Brand"] / device_merge["Registered Users_Total"]
)

device_merge.head()
```

```python
#Q4.4Bar charts 
sample_state = device_merge["State"].iloc[0]

state_device_data = device_merge[
    device_merge["State"] == sample_state
]
plt.figure()
plt.bar(state_device_data["Brand"], state_device_data["Device_Usage_Ratio"])
plt.xticks(rotation=45)
plt.title(f"Device Usage Ratio - {sample_state}")
plt.tight_layout()
plt.show()
```
